```python
import pandas as pd
import os
import kagglehub

dir = kagglehub.dataset_download("marvyaymanhalim/student-performance-and-attendance-dataset")
file_path = os.path.join(dir, "attendance.csv")
df = pd.read_csv(file_path)
df.drop_duplicates(inplace=True)
df['Date'] = pd.to_datetime(df['Date'])
df.columns = df.columns.str.lower()
df.info()
```

```python
df['attendance_status'] = df['attendance_status'].str.strip().str.upper()
df['attendance_status'].value_counts()
```

```python
df_sorted = df.sort_values(by='date')
df_sorted.head()
```

```python
df['attendance_status'] = df['attendance_status'].replace('ABSNT', 'ABSENT')
df['attendance_status'].value_counts()
```

```python
df['status_code'] = (df['attendance_status'] == 'PRESENT').astype(int)
df.info()
```

```python
present_students = df[df['attendance_status'] == 'PRESENT']
present_students.head()
```

```python
df_encoded = pd.get_dummies(df, columns=['subject'])
df_encoded.head()
```
