```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder

# Dataset Creation
data = {
    'Age': [25, 30, 45, np.nan, 50, 22, 35, 61, 28, 42],
    'Salary': [50000, 60000, 120000, 80000, np.nan, 42000, 75000, 250000, 58000, 90000],
    'Gender': ['Male', 'Female', 'Female', 'Male', np.nan, 'Female', 'Male', 'Female', 'Male', 'Female'],
    'City': ['New York', 'London', 'Paris', 'New York', 'London', 'Paris', 'New York', 'London', np.nan, 'Paris'],
    'Date': ['2023-01-15', '2023-02-18', '2023-03-05', '2023-04-12', '2023-05-20', '2023-06-25', '2023-07-30', '2023-08-14', '2023-09-09', '2023-10-01'],
    'Purchase_Amount': [200, 150, 500, 300, 150, 100, 450, 800, 120, 350],
    'Experience': [2, 5, 15, 8, 20, 1, 10, 25, 3, 12],
    'Rating': [4.1, 4.5, 3.8, 4.2, 4.7, 3.5, 4.9, 4.0, 3.9, 4.6]
}
df = pd.DataFrame(data)
display(df)
```

# 🔹 Problem 1: Handling Missing Values

```python
num_cols = df.select_dtypes(include=[np.number]).columns
for col in num_cols:
    df[col] = df[col].fillna(df[col].mean())

cat_cols = df.select_dtypes(include=[object]).columns
for col in cat_cols:
    if col != 'Date':
        df[col] = df[col].fillna(df[col].mode()[0])

display(df)
```

# 🔹 Problem 2: Feature Scaling

```python
print("Before Scaling:")
display(df[['Salary', 'Age']])

scaler_std = StandardScaler()
df['Salary'] = scaler_std.fit_transform(df[['Salary']])

scaler_minmax = MinMaxScaler()
df['Age'] = scaler_minmax.fit_transform(df[['Age']])

print("After Scaling:")
display(df[['Salary', 'Age']])
```

# 🔹 Problem 3: Encoding Categorical Data

```python
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])

df = pd.get_dummies(df, columns=['City'], dtype=int)
display(df)
```

# 🔹 Problem 4: Feature Creation

```python
df['Total_Spending'] = df['Salary'] + df['Purchase_Amount']
display(df)
```

# 🔹 Problem 5: Date Feature Extraction

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Weekday'] = df['Date'].dt.weekday
display(df)
```

# 🔹 Problem 6: Binning

```python
original_age = df['Age'] * (61 - 22) + 22
bins = [0, 25, 45, float('inf')]
labels = ['Young', 'Adult', 'Senior']
df['Age_Group'] = pd.cut(original_age, bins=bins, labels=labels)
display(df)
```

# 🔹 Problem 7: Outlier Detection

```python
q1 = df['Salary'].quantile(0.25)
q3 = df['Salary'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outliers = df[(df['Salary'] < lower_bound) | (df['Salary'] > upper_bound)]
display(outliers)
```

# 🔹 Problem 8: Feature Interaction

```python
df['Experience_Salary'] = df['Experience'] * df['Salary']
display(df)
```

# 🔹 Problem 9: Feature Selection

```python
numeric_df = df.select_dtypes(include=[np.number])
correlations = numeric_df.corr()['Purchase_Amount'].sort_values()
print(correlations)

least_important = correlations.index[0]
if least_important == 'Purchase_Amount':
    least_important = correlations.index[1]

df = df.drop(columns=[least_important])
display(df)
```
