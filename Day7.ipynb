### Plotting a Simple Line Graph
Below, we define a small dataset for monthly sales and use `matplotlib.pyplot` to visualize the trend.

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun' ,'jul']
sales = [150, 180, 210, 190, 240, 280 ,350]

plt.figure(figsize=(8,5))
plt.plot(months, sales, marker='o', color='b', linewidth=2, label='Sales ($)')


plt.title('Monthly Sales Performance', fontsize=14)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Sales', fontsize=12)

plt.show()
```

```python
sales = [150, 180, 210, 190, 240, 280 ,350,100,250,320,180,450,550]
plt.hist(sales,bins=10)
plt.figure(figsize=(8,5))

plt.show()
```

```python
labels = ["A", "B", "C", "D"]
values = [30, 25, 20, 25]
plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Pie Chart")
plt.show()

print("box plot")
data = [12, 15, 14, 10, 18, 20, 15, 13]
plt.boxplot(data)
plt.title("Box Plot")
plt.show()
```

### Temperature Trend using Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 36]

data_df = pd.DataFrame({'Days': days, 'Temperature': temperature})

sns.lineplot(data=data_df, x='Days', y='Temperature', marker='o')
plt.title('Daily Temperature Trend')
plt.xlabel('Days')
plt.ylabel('Temperature')
plt.show()
```
