```python

def is_palindrome(num):
  num_str = str(num)
  return num_str == num_str[::-1]
is_palindrome(145)
```

```python
def sum_digits(num):
  total_sum = 0
  for digit in str(num):
    total_sum += int(digit)
  return total_sum


number_to_sum = 145
result = sum_digits(number_to_sum)
print(f"The sum of digits for {number_to_sum} is: {result}")
```

### Remove duplicates from a list.

```python
list_with_duplicates = [1, 2, 2, 3, 4, 4, 5, 1]
list_without_duplicates = list(set(list_with_duplicates))

print(f"Original list with duplicates: {list_with_duplicates}")
print(f"List without duplicates: {list_without_duplicates}")
```

### Reverse a list without using built-in functions.

```python
original_list = [10, 20, 30, 40, 50]
reversed_list_slicing = original_list[::-1]
reversed_list_manual = []
for i in range(len(original_list) - 1, -1, -1):
  reversed_list_manual.append(original_list[i])

print(f"Original list: {original_list}")
print(f"Reversed list (using slicing): {reversed_list_slicing}")
print(f"Reversed list (using manual loop): {reversed_list_manual}")
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def print_details(self):
        print(f"Name: {self.name}, Age: {self.age}")

s1 = Student("Abhi", 20)
s1.print_details()
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rect = Rectangle(10, 5)
print(rect.area())
print(rect.perimeter())
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def increase_salary(self, amount):
        self.salary += amount

emp = Employee("John", 50000)
emp.increase_salary(5000)
print(emp.salary)
```

```python
class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, item, price):
        self.items[item] = price

    def calculate_total(self):
        return sum(self.items.values())

cart = ShoppingCart()
cart.add_item("Book", 15)
cart.add_item("Pen", 2)
print(cart.calculate_total())
```

```python
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def remove_book(self, book):
        if book in self.books:
            self.books.remove(book)

lib = Library()
lib.add_book("Python Basics")
lib.add_book("Data Science")
lib.remove_book("Data Science")
print(lib.books)
```
