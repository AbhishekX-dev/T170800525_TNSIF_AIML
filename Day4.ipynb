# Day 4 : Numpys and **Pandas**

```python
import numpy as np

marks = np.array([65, 78, 90, 55, 88])
print(marks)
```

```python
average_marks = np.mean(marks)
print(average_marks)
```

```python
highest_marks = np.max(marks)
print(highest_marks)
```

```python
lowest_marks = np.min(marks)
print(lowest_marks)
```

# **Pandas**

```python
import pandas as pd
student_data = {
    'Name': ['Amit', 'Bina', 'Chirag', 'Divya', 'Esha'],
    'Age': [20, 21, 19, 22, 20],
    'Marks': [65, 78, 90, 55, 88]
}
student_df = pd.DataFrame(student_data)
print(student_df)
```

# Display first 5 rows

```python
print("First 5 Rows:")
print(student_df.head())
print("\nDataFrame Info:")
student_df.info()
```

```python
names_col = student_df.loc[:, 'Name']
first_row = student_df.iloc[0]

print("Names Column:\n", names_col)
print("\nFirst Row:\n", first_row)
```

```python
high_scorers = student_df[student_df['Marks'] > 75]
print(high_scorers)
```

```python
null_df = pd.DataFrame({
    'A': [1.0, 2.0, None, 4.0],
    'B': [5.0, None, None, 8.0]
})

print("Original with NaNs:\n", null_df)
print("\nDropped NaNs:\n", null_df.dropna())
print("\nFilled NaNs with 0.0:\n", null_df.fillna(0.0))
```

```python
sorted_students = student_df.sort_values(by='Marks')
print("Sorted by Marks:\n", sorted_students)

student_df['Category'] = ['Batch A', 'Batch B', 'Batch A', 'Batch B', 'Batch A']
group_avg = student_df.groupby('Category')['Marks'].mean()
print("\nAverage Marks by Category:\n", group_avg)
```
